```python
nir=0.72
red=0.30

ndvi=(nir-red)/(nir+red)

print(ndvi)
```

```python
def calculate_ndvi(nir,red):
    (nir-red)/(nir+red)
    return ndvi
```

```python
result=calculate_ndvi(0.72,0.30)
print(result)
```

```python
print(calculate_ndvi(0.80,0.20))
print(calculate_ndvi(0.70,0.20))
print(calculate_ndvi(0.70,0.30))
print(calculate_ndvi(0.80,0.60))
print(calculate_ndvi(0.30,0.10))
```
